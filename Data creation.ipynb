```python
import pandas as pd

# Ruta del archivo Excel
excel_path = 'ruta_a_tu_archivo.xlsx'

# Cargar la hoja 'ISO_Table' del archivo Excel
iso_table_data = pd.read_excel(excel_path, sheet_name='ISO_Table')

# Guardar los datos en formato Parquet usando pandas y pyarrow (o fastparquet)
iso_table_data.to_parquet('iso_table_data.parquet', engine='pyarrow')

print("Archivo Parquet generado exitosamente: 'iso_table_data.parquet'")
```
